```python
bureau_summary = bureau.groupby('SK_ID_CURR').agg(
    total_bureau_records=('SK_ID_BUREAU', 'count'),
    total_credit_sum=('AMT_CREDIT_SUM', 'sum'),
    total_debt=('AMT_CREDIT_SUM_DEBT', 'sum'),
    total_overdue=('AMT_CREDIT_SUM_OVERDUE', 'sum'),
    max_overdue=('AMT_CREDIT_MAX_OVERDUE', 'max')
).reset_index()

bureau_summary.head()
```

```python
prev_summary = prev.groupby('SK_ID_CURR').agg(
    total_previous_apps=('SK_ID_PREV', 'count'),
    avg_prev_credit=('AMT_CREDIT', 'mean'),
    avg_prev_application=('AMT_APPLICATION', 'mean')
).reset_index()

prev_summary.head()
```

```python
customer_master = app[
    [
        'SK_ID_CURR',
        'TARGET',
        'CODE_GENDER',
        'DAYS_BIRTH',
        'AMT_INCOME_TOTAL',
        'AMT_CREDIT',
        'AMT_ANNUITY',
        'EXT_SOURCE_1',
        'EXT_SOURCE_2',
        'EXT_SOURCE_3'
    ]
]
```

```python
analytics_df = (
    customer_master
    .merge(bureau_summary, on='SK_ID_CURR', how='left')
    .merge(prev_summary, on='SK_ID_CURR', how='left')
)
```
